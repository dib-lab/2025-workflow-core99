```python
import glob
import polars as pl
import os
```

```python
depths = glob.glob('../outputs.mapping/cds2/bams/*.depth.txt')
dflist = []
for n, depthfile in enumerate(depths):
    if n and n % 10 == 0:
        print('...', n)
    basename = os.path.basename(depthfile)
    metag, _, sp, _ = basename.split('.', 3)
    #print(metag, sp)
    df = pl.scan_csv(depthfile, separator='\t', has_header=False, new_columns=['contig', 'pos', 'depth'])
    df2 = df.group_by('contig').agg(pl.mean("depth")).with_columns(
        pl.lit(metag).alias("acc"),
        pl.lit(sp).alias("species")
    )
    dflist.append(df2)
```

```python
full_df = pl.concat(dflist)
```

```python
full_df = full_df.collect()
```
